## 🚀 **Operation Next Nest | Student Mini-Projects | Choose Your Adventure**

</div>

---

Pick one of the prompts below. You can work in teams or solo. Keep answers short & visual!

---

### 🏆 **Option A | Design Your Own Scenario**

Create a *new* SAS scenario by changing the weighting of:

- commute time
- housing affordability
- median income
- kids/household demographics
- nearby amenities

**Task:**

1. Add a `%run_scenario(NewScenarioName)` call in SAS
2. Export new CSV from SAS
3. Load into R and add it to the maps
4. Compare against the Base scenario

**Deliverable:**
One slide:
- Top 3 chosen neighborhoods
- A quick “Why your scenario makes sense” note

---

### 🗺 **Option B | Map Makeover (Cartographer Mode)**

Put your own spin on the leaflet maps:

- Add satellite imagery toggle
- Use better icons (🏫 for schools, 🚲 for trails)
- Add highlight/hover effects
- Pan/zoom presets for interesting subregions

**Deliverable:**
Before & after map screenshots + 2 sentences explaining improvements

---

### 🎯 **Option C | “Neighborhood Hall of Fame”**

Find the **top 5 most stable** neighborhoods across scenarios.

**Task:**
Sort by `n_scenarios` in the choropleth and analyze:

- Why does the model love these places?
- Who would love living there?

**Deliverable:**
A short write-up + screenshot of their locations

---

### 🧩 **Option D | Two-Map Story**

Tell a story using:

1. **Weighted score map**
2. **Stability choropleth**

**Prompt Questions:**

- Does the place with the best score always win?
- When might a lower-scoring place be more robust?

**Deliverable:**
Two annotated screenshots + 3–5 bullet insights

---

### 💡 **Option E | Propose a Data Upgrade**

Design a **Phase 2** enhancement:

- School quality?
- Crime/safety?
- Walkability index?
- Climate resilience?
- Broadband access?
- EV charging proximity?

**Deliverable:**
A pitch (5 sentences max)  
Include:
- Why this matters
- Data source you’d use
- How you’d integrate it

---

### 📝 **Submission Suggestion(s)**

- Create a short PDF or slide deck (3–5 pages max)
- Include visuals + labels (no walls of text)

Show not just **what the model did**, but **what YOU learned from it**.
